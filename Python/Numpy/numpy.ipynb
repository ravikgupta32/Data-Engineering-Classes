##Be careful when copying arrays

```python
#Example
a=np.array([1,2,3])
b=a
b[0]=100
print(a)
#Changes the value of a
```

```python
#Write approach to copy the array
a =np.array([1,2,3])
b=a.copy()
b[0]=100
print(a)
```

Reorganizing Arrays

```python
before = np.array([[1,2,3,4],[5,6,7,8]])
print(before)
after = before.reshape((2,2,2))
print(after)
```

```python
#Vertically stacking vectors
v1 = np.array([1,2,3,4])
v2 = np.array([4,5,6,7])
np.vstack([v1,v2,v1,v2])
```

```python
#Horizontal Stack
h1 = np.ones((2,4))
h2 = np.zeros((2,2))
np.hstack((h1,h2))
```
